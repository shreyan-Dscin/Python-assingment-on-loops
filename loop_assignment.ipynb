```python
#args and **kwargs are used in Python to allow functions to accept a variable number of arguments:
#*args is used to pass a variable number of positional arguments.
#**kwargs is used to pass a variable number of keyword arguments.

arg = (11, 26, 34, 40)
total = 0
for num in arg:
    total += num
print("Sum using *args:", total)


kwargs = {"Alice": "Hello", "Bob": "Hi", "Charlie": "Welcome"}
for name, greeting in kwargs.items():
    print(f"{greeting}, {name}!")



```

4)What is a generator function in python? Why yield keyword is used? Give an example of a generator
function.

```python
#A generator function is a special type of function that returns an iterator,
# but instead of using return, it uses the **yield** keyword to produce a sequence of values lazily, one at a time.

#Why is yield used?
#.yield pauses the function and saves its state, allowing it to resume later.
#.It is memory-efficient because it doesn’t store all values in memory, unlike lists.
#.It makes iterating over large sequences faster and more scalable.


gen = (x for x in range(0, 11, 2))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

6)Write a python program to print the first 10 Fibonacci numbers using a while loop.

```python
a, b = 0, 1
count = 0
while count < 10:
    print(a)
    a, b = b, a + b
    count += 1
```

8)Write a python program to check whether a given number is Palindrome or not using a while loop.

```python
num = int(input("Enter a number: "))
original_num = num
reverse_num = 0
while num > 0:
    digit = num % 10
    reverse_num = reverse_num * 10 + digit
    num = num // 10
if original_num == reverse_num:
    print(f"{original_num} is a palindrome.")
else:
    print(f"{original_num} is not a palindrome.")

```

9)Write a code to print odd numbers from 1 to 100 using list comprehension.
Note: Use a list comprehension to create a list from 1 to 100 and use another List comprehension to filter
out odd numbers.

```python
numbers = [num for num in range(1, 101)]
odd_numbers = [num for num in numbers if num % 2 != 0]
print(odd_numbers)
```
